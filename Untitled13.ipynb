```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgba

# 1. 数据
regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
sales = [2354, 1902, 3524, 2698, 2896, 2563]

# 2. 中文字体
plt.rcParams["font.sans-serif"] = [
    "Microsoft YaHei", "SimHei", "PingFang SC",
    "Noto Sans CJK SC", "AR PL UMing CN"
]
plt.rcParams["axes.unicode_minus"] = False

# 3. 创建画布
fig, ax = plt.subplots(figsize=(9, 5.5), dpi=120)
x = np.arange(len(regions))
width = 0.54

# 4. 设置蓝色渐变
bottom = np.array(to_rgba("#0070C0"))
top = np.array(to_rgba("#00B0F0"))

t = np.linspace(0, 1, 256)[:, None]
gradient = (
    bottom[None, :] * (1 - t)
    + top[None, :] * t
).reshape(256, 1, 4)

# 5. 绘制渐变柱形
for xi, value in zip(x, sales):
    ax.imshow(
        gradient,
        extent=[xi-width/2, xi+width/2, 0, value],
        origin="lower",
        aspect="auto",
        zorder=3
    )

    # 添加数值标签
    ax.text(
        xi, value + 70, str(value),
        ha="center", va="bottom", fontsize=11
    )

# 6. 坐标轴与网格
ax.set_xticks(x, regions)
ax.set_xlim(-0.65, len(regions)-0.35)
ax.set_ylim(0, 4000)
ax.set_yticks(np.arange(0, 4001, 1000))
ax.yaxis.grid(True, linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

ax.tick_params(axis="both", length=0)

# 7. 保存图表
plt.tight_layout()
plt.savefig(
    "第一张_渐变柱形图.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import to_rgba

# 1. 数据
regions = ["华北", "华南", "东北", "西北", "西南", "华东"]
sales = [2354, 1902, 3524, 2698, 2896, 2563]

# 2. 中文字体
plt.rcParams["font.sans-serif"] = [
    "Microsoft YaHei", "SimHei", "PingFang SC",
    "Noto Sans CJK SC", "AR PL UMing CN"
]
plt.rcParams["axes.unicode_minus"] = False

# 3. 创建画布
fig, ax = plt.subplots(figsize=(9, 5.5), dpi=120)

# ===== 背景设置 =====
fig.patch.set_facecolor("#F5F9FF")   # 整张图背景
ax.set_facecolor("#EEF5FF")          # 绘图区背景

x = np.arange(len(regions))
width = 0.54

# 4. 设置蓝色渐变
bottom = np.array(to_rgba("#0070C0"))
top = np.array(to_rgba("#00B0F0"))

t = np.linspace(0, 1, 256)[:, None]
gradient = (
    bottom[None, :] * (1 - t)
    + top[None, :] * t
).reshape(256, 1, 4)

# 5. 绘制渐变柱形
for xi, value in zip(x, sales):
    ax.imshow(
        gradient,
        extent=[xi-width/2, xi+width/2, 0, value],
        origin="lower",
        aspect="auto",
        zorder=3
    )

    ax.text(
        xi, value + 70, str(value),
        ha="center", va="bottom", fontsize=11
    )

# 6. 坐标轴与网格
ax.set_xticks(x, regions)
ax.set_xlim(-0.65, len(regions)-0.35)
ax.set_ylim(0, 4000)
ax.set_yticks(np.arange(0, 4001, 1000))
ax.yaxis.grid(True, linestyle="--", alpha=0.3)
ax.set_axisbelow(True)

for spine in ax.spines.values():
    spine.set_visible(False)

ax.tick_params(axis="both", length=0)

# 7. 保存图表
plt.tight_layout()
plt.savefig(
    "第一张_渐变柱形图.png",
    dpi=300,
    bbox_inches="tight",
    facecolor=fig.get_facecolor()
)
plt.show()
```
